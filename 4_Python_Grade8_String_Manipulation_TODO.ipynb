## 6. Cắt chuỗi (slicing)

Cú pháp cơ bản: `text[start:stop]`.
- Lấy từ vị trí `start`.
- Dừng trước vị trí `stop` (không bao gồm `stop`).
- Có thể bỏ `start` hoặc `stop`.
- Có thể dùng bước: `text[start:stop:step]`.

````python
word = "PYTHON"

print(word[0:3])   # P Y T
print(word[2:])    # Từ chỉ số 2 đến hết
print(word[:4])    # Từ đầu đến trước chỉ số 4
print(word[::2])   # Cách một ký tự
print(word[::-1])  # Đảo ngược chuỗi
````

## 14. Kiểm tra loại ký tự

Các phương thức trả về boolean:
- `.isalpha()`: tất cả ký tự là chữ cái và chuỗi không rỗng.
- `.isdigit()`: tất cả ký tự là chữ số và chuỗi không rỗng.
- `.isalnum()`: tất cả ký tự là chữ cái hoặc chữ số và chuỗi không rỗng.
- `.isspace()`: tất cả ký tự là khoảng trắng và chuỗi không rỗng.
- `.islower()`: có ít nhất một ký tự có phân biệt hoa/thường và tất cả ký tự đó là chữ thường.
- `.isupper()`: tương tự với chữ hoa.

````python
print("Python".isalpha())
print("2026".isdigit())
print("Grade8".isalnum())
print("   ".isspace())
print("hello".islower())
print("HELLO".isupper())
print("hello123".isalpha())
````
