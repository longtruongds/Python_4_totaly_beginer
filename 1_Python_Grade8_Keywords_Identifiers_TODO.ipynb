Python không hiểu code như một câu văn thông thường. Một số từ có ý nghĩa cú pháp được quy định sẵn, ví dụ `if` bắt đầu một nhánh điều kiện. Những từ đó là **từ khóa**.

Ngược lại, các tên như `age`, `score`, `student_name` do lập trình viên đặt để gọi biến. Ta được chọn tên phù hợp, miễn tuân thủ quy tắc đặt tên và không trùng từ khóa.

## 1. Năm nhóm cần phân biệt

| Nhóm | Ví dụ | Có thể tự đặt tên không? | Vai trò |
|---|---|---|---|
| **Từ khóa** | `if`, `else`, `and`, `True` | Không | Có ý nghĩa cú pháp/ngữ nghĩa riêng trong Python |
| **Tên biến** | `age`, `score`, `is_ready` | Có | Đặt tên cho dữ liệu |
| **Tên hàm dựng sẵn** | `print`, `input`, `type`, `int` | Không nên ghi đè | Hàm Python có sẵn |
| **Toán tử / dấu cú pháp** | `+`, `==`, `=`, `:` | Không | Thực hiện phép toán hoặc cấu trúc code |
| **Giá trị** | `13`, `3.5`, `"An"`, `True` | Không phải tên biến | Dữ liệu được lưu hoặc xử lý |

Ví dụ: trong `if age >= 13:`, `if` là từ khóa; `age` là tên biến; `>=` là toán tử so sánh; `13` là giá trị.

````python
age = 13

if age >= 13:
    print("Đủ tuổi")

print(type(age))
````

## 2. Từ khóa là gì?

Từ khóa (*keyword*) là từ được Python dành riêng cho một vai trò cụ thể trong ngôn ngữ. Ví dụ:
- `if`, `elif`, `else`: tạo nhánh điều kiện.
- `and`, `or`, `not`: kết hợp hoặc đảo điều kiện logic.
- `True`, `False`: hai giá trị boolean.
- `in`: toán tử kiểm tra thành viên; từ này cũng xuất hiện trong một số cấu trúc khác của Python.

Không được dùng những từ khóa này làm tên biến:

```python
if = 5       # Sai cú pháp
class = "8A" # Sai cú pháp
```

Python sẽ báo `SyntaxError` vì các tên đó đã được dành riêng.

````python
import keyword

print(keyword.iskeyword("if"))
print(keyword.iskeyword("age"))
print(keyword.iskeyword("True"))
print(keyword.iskeyword("print"))
````

## 3. Những từ khóa đã dùng trong các bài trước

Các notebook trước đã sử dụng các từ sau trong điều kiện và biểu thức:

| Từ khóa | Vai trò | Ví dụ |
|---|---|---|
| `if` | Chạy nhánh khi điều kiện đúng | `if age >= 13:` |
| `elif` | Kiểm tra điều kiện tiếp theo nếu nhánh trước không chạy | `elif score >= 8:` |
| `else` | Nhánh còn lại | `else:` |
| `and` | Cả hai điều kiện đều đúng | `age >= 12 and registered` |
| `or` | Ít nhất một điều kiện đúng | `is_member or has_coupon` |
| `not` | Đảo điều kiện | `not has_overdue_book` |
| `in` | Kiểm tra có xuất hiện trong chuỗi/danh sách | `"Py" in "Python"` |
| `True` | Giá trị đúng | `is_registered = True` |
| `False` | Giá trị sai | `homework_done = False` |

### Cần phân biệt
- `True` và `False` là các giá trị boolean được viết bằng từ dành riêng; chúng không phải tên biến thông thường.
- `in` là từ khóa trong Python, đồng thời hoạt động như toán tử kiểm tra thành viên trong biểu thức.
- `type`, `int`, `float`, `str`, `bool`, `print`, `input`, `len`, `round` **không phải từ khóa**. Đây là các tên hàm dựng sẵn.

## 4. Từ khóa khác hàm dựng sẵn như thế nào?

Hàm dựng sẵn (*built-in function*) là hàm Python cung cấp sẵn. Ta gọi hàm bằng cặp ngoặc `(...)`.

Ví dụ:
- `print("Hello")`: gọi hàm `print`.
- `type(13)`: gọi hàm `type`.
- `int("12")`: gọi hàm `int` để chuyển kiểu.

Các tên `print`, `type`, `int`, `str`, `bool`, `input`, `len`, `round` không phải từ khóa. Tuy nhiên, không nên dùng chúng làm tên biến vì có thể che khuất hàm có sẵn và gây lỗi khó hiểu.

````python
import keyword

names = ["if", "else", "and", "True", "type", "int", "print", "input"]

for name in names:
    print(name, "-> keyword?", keyword.iskeyword(name))
````
