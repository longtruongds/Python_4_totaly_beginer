## 5. Xóa dữ liệu: `pop()`, `del`, `clear()`

- `pop(key)`: xóa khóa và trả về giá trị tương ứng.
- `del data[key]`: xóa cặp khóa–giá trị.
- `clear()`: xóa tất cả cặp khóa–giá trị.
- `pop(key, default)`: có thể cung cấp giá trị mặc định nếu khóa không tồn tại.

```python
student = {"name": "Minh", "age": 13, "class_name": "8A"}

removed_age = student.pop("age")
print("Giá trị đã xóa:", removed_age)
print(student)

del student["class_name"]
print(student)

student.clear()
print(student)
```

## 7. Lấy danh sách khóa, giá trị và cặp dữ liệu

- `keys()`: các khóa.

Các phương thức này trả về đối tượng view; khi in ra, có thể thấy dạng `dict_keys(...)`, `dict_values(...)` hoặc `dict_items(...)`.

## 8. Duyệt dictionary bằng `for`

Có thể duyệt khóa, giá trị hoặc cặp khóa–giá trị:
- `for key in data`: duyệt khóa.
- `for value in data.values()`: duyệt giá trị.
- `for key, value in data.items()`: duyệt từng cặp.

```python
student = {"name": "Chi", "age": 13, "score": 9}

for key in student:
    print("Key:", key)

for value in student.values():
    print("Value:", value)

for key, value in student.items():
    print(key, "=>", value)
```

## 14. Đếm số lần xuất hiện bằng dictionary

Dictionary rất hữu ích để đếm số lần mỗi giá trị xuất hiện. Với người mới bắt đầu, có thể dùng `if key in counts` để kiểm tra trước khi tăng bộ đếm.

```python
colors = ["red", "blue", "red", "green", "blue", "red"]
counts = {}

for color in colors:
    if color in counts:
        counts[color] += 1
    else:
        counts[color] = 1

print(counts)
```

## 15. Dictionary và `get()` để đếm gọn hơn

`get(key, default)` giúp lấy giá trị hiện tại hoặc giá trị mặc định nếu khóa chưa tồn tại. Vì vậy có thể viết `counts[item] = counts.get(item, 0) + 1`.

```python
words = ["python", "list", "python", "dict", "list", "python"]
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```
