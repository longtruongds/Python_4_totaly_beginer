## 5. Xóa phần tử: `remove()`, `discard()`, `pop()`, `clear()`

- `remove(value)`: xóa giá trị; nếu không tồn tại sẽ gây `KeyError`.
- `discard(value)`: xóa nếu có; nếu không có thì không báo lỗi.
- `pop()`: xóa và trả về một phần tử bất kỳ; không nên giả định phần tử nào sẽ bị xóa.
- `clear()`: xóa toàn bộ phần tử.

````python
colors = {"red", "blue", "green"}

colors.remove("blue")
print(colors)

colors.discard("purple")  # Không báo lỗi nếu không có
print(colors)

removed_color = colors.pop()
print("Phần tử bị xóa:", removed_color)
print(colors)

colors.clear()
print(colors)
````

## 12. So sánh quan hệ giữa các set

Các phép kiểm tra quan trọng:
- `A.issubset(B)` hoặc `A <= B`: mọi phần tử của A đều thuộc B.
- `A.issuperset(B)` hoặc `A >= B`: A chứa tất cả phần tử của B.
- `A.isdisjoint(B)`: hai set không có phần tử chung.

````python
required_skills = {"variables", "if", "for"}
student_skills = {"variables", "if", "for", "list"}

print(required_skills.issubset(student_skills))
print(student_skills.issuperset(required_skills))
print(required_skills.isdisjoint({"tuple", "dict"}))
````

## 13. Set comprehension

Set comprehension tạo set mới từ một biểu thức và vòng lặp. Có thể kết hợp điều kiện để lọc dữ liệu.

````python
numbers = [1, 2, 2, 3, 4, 4, 5]

squares = {number ** 2 for number in numbers}
even_numbers = {number for number in numbers if number % 2 == 0}

print(squares)
print(even_numbers)
````

## 14. Chuyển đổi giữa Set, List và Tuple

- `set(items)`: tạo set, loại bỏ phần tử trùng lặp.
- `list(my_set)`: chuyển set thành list, nhưng thứ tự không được đảm bảo.
- `tuple(my_set)`: chuyển set thành tuple, thứ tự cũng không được đảm bảo.

Nếu cần giữ thứ tự xuất hiện ban đầu trong khi loại bỏ trùng, có thể dùng `dict.fromkeys()` hoặc một cách duyệt riêng; set đơn thuần không đảm bảo điều đó.

````python
items = ["red", "blue", "red", "green", "blue"]

unique_set = set(items)
unique_list = list(unique_set)
unique_tuple = tuple(unique_set)

print(unique_set)
print(unique_list)
print(unique_tuple)
````
