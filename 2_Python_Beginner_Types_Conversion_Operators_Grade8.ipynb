## 2. Kiểu `int` — số nguyên

Dùng `int` khi làm việc với số nguyên, chẳng hạn số quyển sách hoặc số học sinh. Số nguyên có thể âm, dương hoặc bằng 0.

### 2.1. Các toán tử số học với số nguyên

| Toán tử | Tên | Ví dụ | Kết quả |
|---|---|---|---:|
| `+` | Cộng | `7 + 2` | `9` |
| `-` | Trừ | `7 - 2` | `5` |
| `*` | Nhân | `7 * 2` | `14` |
| `/` | Chia thông thường | `7 / 2` | `3.5` |
| `//` | Chia lấy phần nguyên | `7 // 2` | `3` |
| `%` | Chia lấy dư | `7 % 2` | `1` |
| `**` | Lũy thừa | `7 ** 2` | `49` |

**Điểm cần nhớ:** `/` thường tạo ra kết quả kiểu `float`, kể cả khi phép chia hết.

````python
a = 7
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

print(type(a / b))
````

### 2.2. Chú ý thứ tự thực hiện phép tính

Python tuân theo quy tắc ưu tiên toán tử:
1. Ngoặc `(...)`
2. Lũy thừa `**`
3. Nhân, chia, chia lấy phần nguyên, chia lấy dư: `*`, `/`, `//`, `%`
4. Cộng, trừ: `+`, `-`

Nếu các phép cùng mức ưu tiên, phần lớn được tính từ trái sang phải. Riêng lũy thừa có quy tắc kết hợp khác; ở mức nhập môn, hãy dùng ngoặc để biểu thức dễ đọc.

````python
print(2 + 3 * 4)       # 14: nhân trước, cộng sau
print((2 + 3) * 4)     # 20: tính trong ngoặc trước
print(2 ** 3 + 1)      # 9
````

## 3. Kiểu `float` — số thập phân

Dùng `float` cho các giá trị như chiều cao, thời gian hoặc điểm số có phần thập phân.

### 3.1. Toán tử số học với `float`

Các toán tử `+`, `-`, `*`, `/`, `//`, `%`, `**` cũng dùng được với số thực. Kết quả có thể có nhiều chữ số thập phân.

````python
price = 12.5
quantity = 3

print(price + 1.5)
print(price - 2.0)
print(price * quantity)
print(price / 2)
print(7.5 // 2)
print(7.5 % 2)
print(1.5 ** 2)
````

### 3.2. Làm tròn và sai số số thực

````python
print(round(3.14159, 2))
print(round(12.567, 1))
print(0.1 + 0.2)
````

## 4. Kiểu `str` — chuỗi ký tự

`str` dùng để lưu văn bản. Chuỗi có thể chứa chữ, số, dấu cách và ký hiệu.

Ví dụ:
- `"Hello"`
- `"Lớp 8"`
- `"2026"` — đây là văn bản, không phải số nguyên.

### 4.1. Các toán tử với chuỗi

| Toán tử | Tác dụng | Ví dụ |
|---|---|---|
| `+` | Nối chuỗi | `"Xin " + "chào"` |
| `*` | Lặp chuỗi | `"Ha" * 3` |
| `in` | Kiểm tra chuỗi con có xuất hiện không | `"Py" in "Python"` |
| `not in` | Kiểm tra không xuất hiện | `"Java" not in "Python"` |

Không thể dùng phép trừ hoặc chia trực tiếp giữa hai chuỗi.

````python
first_name = "Minh"
greeting = "Xin chào, "

print(greeting + first_name)
print("Ha" * 3)
print("Py" in "Python")
print("Java" not in "Python")
````

### 4.2. Một số thao tác hữu ích với chuỗi

- `len(text)`: đếm số ký tự.
- `text.lower()`: chuyển chữ thành chữ thường.
- `text.upper()`: chuyển chữ thành chữ hoa.
- `text.strip()`: bỏ khoảng trắng ở đầu và cuối.
- `text[0]`: lấy ký tự đầu tiên; vị trí bắt đầu từ `0`.

Các thao tác này không phải toán tử, nhưng thường được dùng cùng chuỗi.

````python
message = "  Hello Python  "

print(len(message))
print(message.lower())
print(message.upper())
print(message.strip())
print(message[0])       # Ký tự đầu tiên thực tế là dấu cách
print("Python"[0])
print("Python"[1:4])    # Lấy các ký tự ở vị trí 1, 2, 3
````

### 4.3. Ghép văn bản với số

Đoạn code sau gây lỗi vì Python không tự động nối `str` và `int` bằng `+`:

## 5. Kiểu `bool` — đúng hoặc sai

`bool` chỉ có hai giá trị:
- `True`: đúng.
- `False`: sai.

Lưu ý viết hoa chữ cái đầu. `true` và `false` viết thường không phải các giá trị boolean hợp lệ trong Python.

Boolean thường là kết quả của phép so sánh hoặc phép logic.

````python
print(type(True))
print(type(False))

print(5 > 3)
print(5 < 3)
print(5 == 5)
print(5 != 5)
````

### 5.1. Toán tử so sánh

| Toán tử | Ý nghĩa |
|---|---|
| `==` | Bằng nhau |
| `!=` | Khác nhau |
| `>` | Lớn hơn |
| `<` | Nhỏ hơn |
| `>=` | Lớn hơn hoặc bằng |
| `<=` | Nhỏ hơn hoặc bằng |

**Phân biệt rất quan trọng:**
- `=` là phép gán giá trị cho biến.
- `==` là phép so sánh hai giá trị.

Kết quả của phép so sánh là `True` hoặc `False`.

````python
score = 8

print(score >= 5)
print(score == 10)
print(score != 0)
print(score < 9)
print(score <= 8)
````

### 5.2. Toán tử logic `and`, `or`, `not`

- `A and B`: đúng khi **cả A và B đều đúng**.
- `A or B`: đúng khi **ít nhất một trong A hoặc B đúng**.
- `not A`: đảo ngược giá trị đúng/sai của A.

Ví dụ: để được tham gia câu lạc bộ, học sinh phải từ 13 tuổi trở lên **và** đã đăng ký.

````python
age = 13
registered = True

print(age >= 13 and registered)
print(age < 13 or registered)
print(not registered)
````

## 6. Chuyển đổi kiểu dữ liệu (Type conversion)

Chuyển đổi kiểu nghĩa là biến một giá trị từ kiểu này sang kiểu khác.

| Hàm | Ví dụ | Kết quả |
|---|---|---|
| `int(x)` | `int("12")` | `12` |
| `float(x)` | `float("3.5")` | `3.5` |
| `str(x)` | `str(12)` | `"12"` |
| `bool(x)` | `bool(1)` | `True` |

Chuyển đổi chỉ thành công khi giá trị phù hợp với kiểu đích. Ví dụ `int("hello")` gây `ValueError` vì `"hello"` không phải văn bản biểu diễn một số nguyên hợp lệ.

````python
print(int("12"))
print(float("3.5"))
print(str(12))
print(bool(1))

print(type(int("12")))
print(type(float("3.5")))
print(type(str(12)))
````

### 6.1. Chuyển đổi giữa `int` và `float`

- `float(5)` tạo ra `5.0`.
- `int(5.9)` cho ra `5`: phần thập phân bị bỏ, không phải làm tròn.
- `round(5.9)` làm tròn theo quy tắc của Python; với ví dụ này kết quả là `6`.

Đừng nhầm `int()` với `round()`.

````python
print(float(5))
print(int(5.9))
print(int(-2.9))
print(round(5.9))
print(round(5.2))
````

### 6.2. Chuyển đổi văn bản thành số

### 6.3. `bool()` và giá trị truthy/falsy

Đây là điểm dễ gây nhầm: `bool("False")` không phân tích chữ `"False"` theo nghĩa tiếng Việt; nó chỉ kiểm tra chuỗi có rỗng hay không.

````python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Hello"))
print(bool("False"))
````

## 12. Bảng ghi nhớ nhanh

| Chủ đề | Cú pháp | Ghi nhớ |
|---|---|---|
| Kiểm tra kiểu | `type(x)` | Cho biết kiểu của giá trị |
| Số nguyên | `int` | Không có phần thập phân |
| Số thực | `float` | Có thể có phần thập phân |
| Chuỗi | `str` | Văn bản trong dấu nháy |
| Đúng/sai | `bool` | `True` hoặc `False` |
| Chuyển sang số nguyên | `int(x)` | `int(5.9)` thành `5` |
| Chuyển sang số thực | `float(x)` | `float("2.5")` thành `2.5` |
| Chuyển sang chuỗi | `str(x)` | Hữu ích khi ghép văn bản |
| Cộng / trừ / nhân | `+ - *` | Toán tử số học |
| Chia / chia nguyên / dư | `/ // %` | Phân biệt rõ |
| Lũy thừa | `**` | Ví dụ `2 ** 3` là `8` |
| So sánh | `== != > < >= <=` | Trả về boolean |
| Logic | `and or not` | Kết hợp điều kiện |
| Nối chuỗi | `"a" + "b"` | Cả hai toán hạng là chuỗi |
| Lặp chuỗi | `"ha" * 3` | Tạo `"hahaha"` |
| Kiểm tra thành viên | `in`, `not in` | Kiểm tra có/không xuất hiện |
| Gán kết hợp | `+=`, `-=`, `*=`... | Cập nhật biến ngắn gọn |
