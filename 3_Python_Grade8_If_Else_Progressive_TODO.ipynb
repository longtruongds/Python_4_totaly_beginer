## 4. `if / elif / else`: nhiều khả năng

Khi có nhiều trường hợp, dùng `elif` (viết tắt của *else if*). Python kiểm tra từ trên xuống dưới và chạy nhánh đầu tiên có điều kiện đúng.

Ví dụ phân loại điểm:
- Từ 8 trở lên: tốt.
- Từ 6.5 đến dưới 8: khá.
- Từ 5 đến dưới 6.5: đạt.
- Dưới 5: cần cố gắng.

````python
score = 7.2

if score >= 8:
    print("Tốt")
elif score >= 6.5:
    print("Khá")
elif score >= 5:
    print("Đạt")
else:
    print("Cần cố gắng")
````

**Vì sao không cần viết `score < 8` ở nhánh thứ hai?**

Nếu chương trình đã đến `elif score >= 6.5`, điều đó có nghĩa nhánh trước `score >= 8` đã sai. Vì vậy, `score` chắc chắn nhỏ hơn 8. Đây là một cách viết gọn, nhưng cần hiểu thứ tự kiểm tra.

## 16. Bảng ghi nhớ

| Cấu trúc | Dùng khi nào? |
|---|---|
| `if condition:` | Thực hiện lệnh khi điều kiện đúng |
| `if ... else ...` | Chọn một trong hai nhánh |
| `if ... elif ... else ...` | Chọn giữa nhiều trường hợp |
| `and` | Tất cả điều kiện phải đúng |
| `or` | Ít nhất một điều kiện đúng |
| `not` | Đảo giá trị đúng/sai |
| `in` | Kiểm tra phần tử/chuỗi con có xuất hiện |
| `==` | So sánh bằng |
| `=` | Gán giá trị |
| `>=` | Lớn hơn hoặc bằng |
