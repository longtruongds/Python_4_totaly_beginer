## 2. Từ khóa là gì?

**Từ khóa (*keyword*)** là từ được Python dành riêng cho một vai trò trong cú pháp hoặc ngữ nghĩa. Ví dụ:
- `if`, `elif`, `else`: rẽ nhánh.
- `for`, `while`, `break`, `continue`: vòng lặp.
- `def`, `return`, `lambda`: hàm.
- `class`: định nghĩa lớp.
- `True`, `False`, `None`: giá trị đặc biệt.
- `and`, `or`, `not`: logic.

Không được dùng từ khóa làm tên biến:

```python
if = 5       # SyntaxError
class = "8A" # SyntaxError
```

Tên như `if_score`, `class_name`, `print`, `type`, `int` không phải từ khóa. Tuy nhiên, tránh dùng tên hàm dựng sẵn như `print` làm tên biến vì sẽ che khuất hàm có sẵn.

## 3. Danh mục từ khóa theo nhóm chức năng

Danh sách dưới đây được chia theo nhóm để dễ học. Danh sách chính xác của phiên bản hiện tại luôn có thể được lấy bằng `keyword.kwlist`.

## 3.1. Điều kiện và logic

- **`if`** — Bắt đầu nhánh khi điều kiện đúng.
- **`elif`** — Kiểm tra điều kiện tiếp theo khi nhánh trước không được chọn.
- **`else`** — Nhánh còn lại khi các điều kiện trước đều sai.
- **`and`** — Đúng khi cả hai điều kiện đều đúng.
- **`or`** — Đúng khi ít nhất một điều kiện đúng.
- **`not`** — Đảo giá trị đúng/sai.
- **`in`** — Kiểm tra thành viên/phần tử; cũng xuất hiện trong cấu trúc cú pháp.
- **`is`** — Kiểm tra hai tham chiếu có cùng trỏ đến một đối tượng hay không; thường dùng với `None`.

## 3.2. Vòng lặp

- **`for`** — Duyệt qua các phần tử của một đối tượng có thể lặp.
- **`while`** — Lặp khi điều kiện còn đúng.
- **`break`** — Kết thúc vòng lặp gần nhất.
- **`continue`** — Bỏ qua phần còn lại của lượt lặp hiện tại.
- **`pass`** — Câu lệnh không làm gì; dùng làm chỗ giữ chỗ.

## 3.3. Hàm và phạm vi biến

- **`def`** — Bắt đầu định nghĩa hàm.
- **`return`** — Kết thúc hàm và trả kết quả.
- **`lambda`** — Tạo hàm nhỏ dạng biểu thức.
- **`yield`** — Tạm dừng generator và phát ra một giá trị.
- **`global`** — Khai báo tên thuộc phạm vi toàn cục của module.
- **`nonlocal`** — Tham chiếu biến ở phạm vi hàm bao ngoài gần nhất.

## 3.4. Lớp và đối tượng

- **`class`** — Bắt đầu định nghĩa một lớp.

## 3.5. Ngoại lệ và quản lý tài nguyên

- **`try`** — Bắt đầu khối có thể phát sinh ngoại lệ.
- **`except`** — Xử lý ngoại lệ phù hợp.
- **`finally`** — Khối thường được chạy dù có ngoại lệ hay không.
- **`raise`** — Chủ động phát sinh ngoại lệ.
- **`assert`** — Kiểm tra điều kiện; nếu sai phát sinh `AssertionError`.
- **`with`** — Dùng context manager để quản lý tài nguyên/trạng thái.
- **`as`** — Đặt bí danh hoặc gán đối tượng trong cấu trúc như `with`, `except`, `import`.

## 3.6. Module và tên

- **`import`** — Nạp module hoặc tên.
- **`from`** — Chỉ định module nguồn khi nhập một phần module.
- **`del`** — Xóa tên hoặc phần tử khỏi cấu trúc phù hợp.

## 3.7. Giá trị đặc biệt

- **`True`** — Giá trị boolean đúng.
- **`False`** — Giá trị boolean sai.
- **`None`** — Giá trị đặc biệt biểu thị không có giá trị hoặc chưa có kết quả.

## 3.8. Bất đồng bộ

- **`async`** — Khai báo hàm hoặc cấu trúc bất đồng bộ.
- **`await`** — Chờ kết quả awaitable trong hàm bất đồng bộ.

## 4. Ví dụ kết hợp từ khóa

`def`, `if`, `elif`, `else`, `return` và `for` có thể phối hợp trong cùng một chương trình.

````python
def classify_score(score):
    if score >= 8:
        return "Giỏi"
    elif score >= 5:
        return "Đạt"
    else:
        return "Cần cố gắng"

for score in [9, 7, 4]:
    print(score, classify_score(score))
````

## 5. Từ khóa mềm (*soft keywords*)

Một số từ chỉ có ý nghĩa đặc biệt trong một số ngữ cảnh. Trong các phiên bản Python hiện đại, các ví dụ có thể gồm `match`, `case`, `_` và `type` (trong cú pháp khai báo kiểu mới ở Python 3.12 trở lên).

Danh sách chính xác phụ thuộc phiên bản. Dùng `keyword.softkwlist` để kiểm tra.

````python
import keyword

soft_keywords = getattr(keyword, "softkwlist", [])
print("Soft keywords:", soft_keywords)

for word in soft_keywords:
    print(word, "hard keyword?", keyword.iskeyword(word))
````

### Ví dụ `match/case`

Cú pháp so khớp mẫu `match/case` có từ Python 3.10 trở lên. Nếu dùng phiên bản cũ hơn, ô này có thể không chạy.

````python
command = "start"

match command:
    case "start":
        print("Bắt đầu")
    case "stop":
        print("Dừng lại")
    case _:
        print("Lệnh không được nhận diện")
````

## 6. Cách tự kiểm tra từ khóa

Module chuẩn `keyword` cung cấp:
- `keyword.iskeyword(name)`: kiểm tra từ khóa cứng.
- `keyword.kwlist`: danh sách từ khóa cứng.
- `keyword.issoftkeyword(name)`: kiểm tra từ khóa mềm trên phiên bản hỗ trợ.
- `keyword.softkwlist`: danh sách từ khóa mềm trên phiên bản hỗ trợ.

Đây là cách đáng tin cậy hơn danh sách không ghi rõ phiên bản.

````python
import keyword

words = ["if", "else", "and", "True", "type", "print", "match", "case"]
soft_check = getattr(keyword, "issoftkeyword", lambda name: False)

for word in words:
    print(
        f"{word:>12} | keyword={keyword.iskeyword(word)!s:<5} "
        f"| soft_keyword={soft_check(word)}"
    )
````
