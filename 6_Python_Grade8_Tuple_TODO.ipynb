## 3. Truy cập phần tử bằng index

Chỉ số của tuple bắt đầu từ `0`:
- `items[0]`: phần tử đầu tiên.
- `items[1]`: phần tử thứ hai.
- `items[-1]`: phần tử cuối cùng.
- `items[-2]`: phần tử áp chót.

Truy cập một index không tồn tại sẽ gây `IndexError`.

````python
colors = ("red", "blue", "green", "yellow")

print(colors[0])
print(colors[2])
print(colors[-1])
print(colors[-2])
````

## 4. Slicing — lấy một phần tuple

Cú pháp `items[start:stop]` lấy phần tử từ `start` đến trước `stop`.
- `items[:3]`: ba phần tử đầu.
- `items[2:]`: từ index 2 đến hết.
- `items[1:4]`: index 1, 2, 3.
- `items[::-1]`: tuple mới theo thứ tự đảo ngược.

````python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[:3])
print(numbers[2:])
print(numbers[1:4])
print(numbers[::-1])
````

## 8. Dùng `count()` và `index()`

Tuple có hai phương thức tra cứu thường dùng:
- `items.count(value)`: đếm số lần giá trị xuất hiện.
- `items.index(value)`: trả về vị trí xuất hiện đầu tiên; nếu không có giá trị sẽ phát sinh `ValueError`.

````python
votes = ("red", "blue", "red", "green", "red")

print(votes.count("red"))
print(votes.index("green"))
````

## 12. Chuyển đổi giữa list và tuple

- `tuple(my_list)`: chuyển list thành tuple.
- `list(my_tuple)`: chuyển tuple thành list.

Có thể chuyển đổi khi cần sửa dữ liệu rồi tạo lại tuple. Tuy nhiên, hãy hiểu rõ rằng list và tuple có hành vi khác nhau.

````python
colors_list = ["red", "green", "blue"]
colors_tuple = tuple(colors_list)

back_to_list = list(colors_tuple)
back_to_list.append("yellow")

print(colors_list)
print(colors_tuple)
print(back_to_list)
````

## 13. List và tuple: khác nhau ở đâu?

| Đặc điểm | List | Tuple |
|---|---|---|
| Cú pháp thường gặp | `[1, 2, 3]` | `(1, 2, 3)` |
| Có thứ tự | Có | Có |
| Cho phép trùng phần tử | Có | Có |
| Sửa trực tiếp phần tử | Có | Không |
| Thêm/xóa bằng phương thức | Có | Không |
| Phù hợp khi | Dữ liệu cần thay đổi | Dữ liệu nên giữ cố định |

Lưu ý: tuple có thể chứa một list; khi đó list bên trong vẫn có thể thay đổi. Tính bất biến của tuple không tự động làm mọi đối tượng bên trong trở thành bất biến.

````python
a_list = [1, 2, 3]
a_tuple = (1, 2, 3)

a_list[0] = 99
print(a_list)

# Không chạy dòng sau vì nó gây TypeError:
# a_tuple[0] = 99
print(a_tuple)
````

## 17. Bảng tra cứu nhanh
