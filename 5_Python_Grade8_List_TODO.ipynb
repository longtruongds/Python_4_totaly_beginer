## 3. Slicing — lấy một phần danh sách

Cú pháp `items[start:stop]` lấy phần tử từ `start` đến trước `stop`.
- `items[:3]`: ba phần tử đầu.
- `items[2:]`: từ chỉ số 2 đến hết.
- `items[1:4]`: chỉ số 1, 2, 3.
- `items[::-1]`: bản sao theo thứ tự đảo ngược.

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[:3])
print(numbers[2:])
print(numbers[1:4])
print(numbers[::-1])
````

## 6. Xóa phần tử: `remove()`, `pop()`, `del`

- `remove(value)`: xóa lần xuất hiện đầu tiên của giá trị.
- `pop(index)`: xóa và trả về phần tử tại index; nếu bỏ index thì xóa phần tử cuối.
- `del items[index]`: xóa phần tử tại index.
- `clear()`: xóa tất cả phần tử.

Lưu ý: `remove()` báo `ValueError` nếu giá trị không tồn tại.

````python
colors = ["red", "blue", "green", "blue"]

colors.remove("blue")
print(colors)

removed_color = colors.pop()
print("Đã lấy ra:", removed_color)
print(colors)

del colors[0]
print(colors)

colors.clear()
print(colors)
````

## 10. Tính toán với list số

Một số hàm dựng sẵn hữu ích:
- `sum(numbers)`: tổng các phần tử số.
- `min(numbers)`: giá trị nhỏ nhất.
- `max(numbers)`: giá trị lớn nhất.
- `len(numbers)`: số phần tử.
- Trung bình có thể tính bằng `sum(numbers) / len(numbers)` nếu list không rỗng.

````python
scores = [8, 9, 7, 10, 6]

print("Tổng:", sum(scores))
print("Nhỏ nhất:", min(scores))
print("Lớn nhất:", max(scores))
print("Số điểm:", len(scores))
print("Trung bình:", sum(scores) / len(scores))
````

## 12. Sắp xếp và đảo thứ tự

- `items.sort()`: sắp xếp list tại chỗ, làm thay đổi list ban đầu.
- `items.sort(reverse=True)`: sắp xếp giảm dần.
- `sorted(items)`: trả về list mới đã sắp xếp, không thay đổi list gốc.
- `items.reverse()`: đảo thứ tự tại chỗ, không phải sắp xếp.

````python
numbers = [5, 2, 9, 1, 7]

print(sorted(numbers))
print(numbers)

numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)

numbers.reverse()
print(numbers)
````

## 14. Copy list — một lỗi thường gặp

Nếu viết `b = a`, cả hai tên cùng tham chiếu đến cùng một list. Thay đổi qua `b` cũng ảnh hưởng đến `a`.

Dùng `a.copy()` hoặc `a[:]` để tạo bản sao nông (*shallow copy*) cho list đơn giản.

````python
a = [1, 2, 3]
b = a
b.append(4)

print("a:", a)
print("b:", b)

c = a.copy()
c.append(5)

print("a sau khi sửa c:", a)
print("c:", c)
````

## 19. Bảng tra cứu nhanh

| Mục đích | Cú pháp |

| Tạo list | `[1, 2, 3]` |

| Thêm cuối | `append(value)` |
| Chèn vị trí | `insert(index, value)` |

| Xóa và lấy phần tử | `pop(index)` |
| Xóa theo index | `del items[index]` |
| Xóa toàn bộ | `clear()` |
| Đếm phần tử | `len(items)` |
| Tìm kiếm | `value in items` |
| Đếm lần xuất hiện | `count(value)` |
| Tìm index | `index(value)` |
| Sắp xếp tại chỗ | `sort()` |
| Sắp xếp tạo list mới | `sorted(items)` |
| Đảo thứ tự | `reverse()` |
| Sao chép | `copy()` |
| Tổng/nhỏ nhất/lớn nhất | `sum()`, `min()`, `max()` |
